```python
try:    
    num_list = [2, 6, 34, 65, 11]
    index = int(input("인덱스 입력 : "))
    print(num_list[index])
except (ValueError, IndexError):
    print(f"0부터 {len(num_list)-1} 까지의 정수를 입력하세요.")
except Exception as e:
    print(f"예외가 발생했습니다 : {e} -> {type(e)}")
```

```python
try:
    number = int(input("숫자 입력 : "))
except ValueError:
    print("숫자를 입력하지 않았습니다.")
else:
    print(number)
finally:
    print("끝났습니다.")
```

```python
while True:
    try:
        number = int(input("양의 정수 입력 : "))
        if number <= 0:
            raise ValueError
    except ValueError:
        print("양의 정수를 입력하세요.")
    else:
        sum = 0
        for i in range(1, number+1):
            sum += i
        print(f">> 1부터 {number}까지의 합 : {sum}")
        break
```

### 학습 내용 되돌아보기 (2)

```python
try:
    m = int(input("월(month) 입력 > "))
    if not (1 <= m <= 12):
        raise Exception("1~12 사이의 정수를 입력하세요.")   
except Exception as e:
    print(f"예외 발생: {e}")
else:
    print(f"{m}월: 잘 입력했습니다.")

# 5 -> 5월: 잘 입력했습니다.
# 0 -> 예외 발생: 1~12 사이의 정수를 입력하세요.
```

### 학습 내용 되돌아보기 (3)

```python
def divide(x, y):
    try:
        x / y
    except:
        print("0으로는 나눌 수 없습니다.")
    else:
        print(f"결과: {x/y}")

divide(3.2, 2)
divide(5.4, 0)
```

### 학습 내용 되돌아보기 (4)

```python
uyear = {1: "freshman", 2: "sophomore", 3: "junior", 4: "senior"}

try:
    y = int(input("대학교 몇 학년이지요? "))

    if not (1 <= y <= 4):
        raise Exception
except Exception as e:
    print(f"예외 발생 이름: {type(e)}")
    print(f"예외 발생 이유: 1-4 정수를 입력하세요.")
else:
    print(f"{y}학년: {uyear[y]}")
finally:
    print("예외 처리가 잘되는군요!")
```
