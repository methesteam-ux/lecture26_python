### 딕셔너리 만들기

```python
dic = {
    "name" : "pey", 
    "phone" : "010-9999-1234", 
    "birth" : "1118"
}
print(dic)

a = {1: "hi"}
print(a)
a = {"a": [1, 2, 3]}
print(a)

print()

# 체스판의 말 위치 저장
chess_board = {
    (0, 0): "White Rook", 
    (0, 1): "White Knight",
    (7, 7): "Black Rook"
}
print(chess_board)

print()

# 딕셔너리에서 키는 유일해야 함, 만약 키가 동일한 항목이 여러 개 있다면 마지막 항목을 제외한 나머지 항목들은 무시됨
prices = {"아메리카노": 2000, "라떼": 3000, "아메리카노": 3500, "녹차": 2500}
print(prices)

print()

# 비어있는 딕셔너리의 생성
print({})
print(dict())
```

### 딕셔너리의 항목에 접근하기

```python
coffee_menu = {"아메리카노": 2000, "라떼": 3000, "녹차": 2500}
print(coffee_menu["라떼"])

try:
    print(coffee_menu["카푸치노"])
except:
    print("오류 발생")

print(coffee_menu.get("카푸치노", "없는메뉴"))

print("카푸치노" in coffee_menu)

print()

dict_items = coffee_menu.items()
print(dict_items)
dict_keys = coffee_menu.keys()
print(dict_keys)
dict_values = coffee_menu.values()
print(dict_values)

for i in coffee_menu:
    print(i)

print()

# 딕셔너리의 항목 수정, 추가하기
coffee_menu["라떼"] = 3500
coffee_menu["카푸치노"] = 4000
print(coffee_menu)

print()

# 딕셔너리의 항목 삭제하기
data = coffee_menu.pop("녹차")
print(data)
print(coffee_menu)
del coffee_menu["라떼"]
print(coffee_menu)
coffee_menu.clear()
print(coffee_menu)
```

### [실습] 모스 부호 번역기

```python
codes = {
    "a": ".-", 
    "b": "-...", 
    "c": "-.-.", 
    "d": "-..", 
    "e": ".", 
    "f": "..-.", 
    "g": "--.", 
    "h": "....", 
    "i": "..", 
    "j": ".---", 
    "k": "-.-", 
    "l": ".-..", 
    "m": "--", 
    "n": "-.", 
    "o": "---", 
    "p": ".--.", 
    "q": "--.-", 
    "r": ".-.", 
    "s": "...", 
    "t": "-", 
    "u": "..-", 
    "v": "...-", 
    "w": ".--", 
    "x": "-..-", 
    "y": "-.--", 
    "z": "--.."
}

text = input("> 알파벳 문자열 입력 :")

def translate(a):
    if len(a) == 1 and a.isalpha():
        return codes[a]
    else:
        print("변환 오류 : 함수에 입력값이 여러개이거나 문자가 아닙니다.")
        return -1

if not text.isalpha():
    print("알파벳 문자열이 아닙니다.")
else:
    mores = ""
    for i in range(len(text)):
        mores = mores + translate(text[i]) + " "

    print(mores.rstrip())
```
