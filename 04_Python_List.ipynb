### 리스트(List) 만들기

```python
a = []
b = list()
c = [1, 2, 3]
d = ["banana", "apple", "tomato", "watermelon"]
e = [1, 2, 3, "one", "two", "three"]
f = [1, 2, ["banana", 100], ["tomato", 300]]

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

### 리스트의 항목에 접근하기 - 인덱싱(indexing), for문으로 접근하기

```python
test_list = [1, 2, 3, "one", "two", "three"]

print(test_list[0])
print(test_list[3])
print(test_list[-1])
print(test_list[-3])

print()

for item in test_list:
    print(item)
```

### 리스트의 항목에 접근하기 - 슬라이싱(slicing)

```python
test_list = [1, 2, 3, 4, 5, 6, 7, 8]

print(test_list[0:5])
print(test_list[:5])

print()

print(test_list[1::3])
print(test_list[:5:2])
print(test_list[::3])
print(test_list[:])
print(test_list[::])

print()

print(test_list[-1:-7:-2])
print(test_list[1:-1:])
print(test_list[-3::-1])
print(test_list[::-1])
print(test_list[::-2])

print()

test_list[:2] = [111, 222, 333]
print(test_list)
```

### 리스트에 항목 추가하기

```python
a_list = [1, 2, 3, 4]
a_list.append("four")
a_list.insert(3, "three")
a_list.extend([10, 11])
print(a_list)

print()

b_list = [1, 2, 3]
c_list = ["a", "b", "c"]
d_list = b_list + c_list
print(d_list)

print()

e_list = b_list * 3
print(e_list)
```

### 리스트에서 항목 삭제하기

```python
test_list = [1, 2, 3, "a", "b", "c", 10, 11]

item = test_list.pop(2)
print(item)
item = test_list.pop()
print(item)
test_list.remove("a")
print(test_list)

print()

del test_list[2]
print(test_list)
del test_list[:2]
print(test_list)

print()

test_list.clear()
print(test_list)
```

### 리스트의 항목 재배열하기

```python
test_list = [1, 2, 3, 4, 5, 6, 7]
test_list.reverse()
print(test_list)

print()

a_list = ["one", "two", "three", "four"]
print("원본 데이터 :", a_list)
a_list.reverse()
print("역순 후 데이터 :", a_list)
a_list.sort()
print("정렬 후 데이터 :", a_list)

print()

a_list = ["one", "two", "three", "four"]
b_list = sorted(a_list)
print("원본 데이터 :", a_list)
print("정렬 후 데이터 :", b_list)

print()

c_list = sorted(a_list, key=len)
print(c_list)
```

```python
scores = [95, 45, 21, 74, 55]
print(sorted(scores, reverse=True))
print(scores)

scores.sort(reverse=True)
print(scores)
```

### 리스트 탐색하기

```python
test_list = [1, 2, 3, 1, 2, 3]

target = int(input("숫자 :"))

print(target in test_list)
print(test_list.index(target))
print(test_list.count(target))
print(len(test_list))
print(max(test_list))
print(min(test_list))
print(sum(test_list))
```

```python
# 1~N의 수로 이루어진 리스트 생성
N = int(input("숫자 :"))
num_list = []

for i in range(1, N+1):
    num_list.append(i)

print(num_list)
```

### 리스트 내포, list comprehension

```python
# 1~N의 수로 이루어진 리스트 생성
N = int(input("숫자 :"))

num_list = list(range(1, N+1))
print(num_list)

print()

num_list = [i for i in range(1, N+1)]
print(num_list)

print()

# expression
num_list = [i*2 for i in range(1, N+1)]
print(num_list)

print()

# if문 사용
num_list = [i for i in range(1, N+1) if i%2 == 0]
print(num_list)
```
