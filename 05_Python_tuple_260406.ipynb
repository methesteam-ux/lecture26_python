### 튜플(tuple)의 개념

```python
t1 = (1, 2, 3)
t2 = ("a", "b", ("ab", "cd"))
t3 = ("tuple", (1, 2), [3, 4])

try:
    t1.append(4)
except:
    print("튜플에 값 추가 불가")

try:
    t2.remove("a")
except:
    print("튜플 값 제거 불가")

t3[2].append(5)
print(t3)
```

### 튜플(tuple) 만들기

```python
t4 = ()
t5 = (1, )
print(t4)
print(t5)

print()

t6 = 1, 2, 3
t7 = 1,
print(t6)
print(t7)

print()

x, y, z = t6
print(x, y, z)
```

### 튜플의 활용

```python
x, y = 10, 20
x, y = y, x
print(x, y)

print()

def add_and_mul(a, b):
    return a+b, a*b

result = add_and_mul(3, 5)
print(result)
```
