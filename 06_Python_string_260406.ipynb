### 문자열을 원하는 형식으로 만들기 : f-문자열 (Formatted String)

```python
print(f"f-string은 {{변수명}}과 같이 사용합니다.")

# 산술 연산
print(f"2 + 3 = {2+3}")

# 메서드 호출
lecture = "python"
print(f"Hello, {lecture.upper()}!")

# 조건문 (삼항 연산자)
n = 10
print(f"{n}은 {"짝수" if n%2 == 0 else "홀수"}입니다.")

number = 1234567.89123
print(f"소수점 자릿수 제한: {number:.2f}")
print(f"콤마 추가: {number:,}")
print(f"동시 사용: {number:,.2f}")
```

### [참고] format() 메서드, % 서식 지정

```python
# 위치 기반
print("I like {} and {}.".format("Python", "Java"))

# 인덱스 지정 (순서 변경 가능)
print("{1} comes after {0}.".format("A", "B"))

# 이름 지정 (키워드 인자)
print("{fruit} is red.".format(fruit="Apple"))

# % 서식 지정
name = "Alice"
age = 25
print("Name: %s, Age: %d" % (name, age))
```

### [실습] 알파벳 두 개를 입력 받아서 두 개 사이에 몇 개의 알파벳이 있는지 계산

### [실습] 모스 부호 번역기

```python
codes = [
    ".-", 
    "-...", 
    "-.-.", 
    "-..", 
    ".", 
    "..-.", 
    "--.", 
    "....", 
    "..", 
    ".---", 
    "-.-", 
    ".-..", 
    "--", 
    "-.", 
    "---", 
    ".--.", 
    "--.-", 
    ".-.", 
    "...", 
    "-", 
    "..-", 
    "...-", 
    ".--", 
    "-..-", 
    "-.--", 
    "--.."
]

text = input("> 알파벳 문자열 입력 :")

def translate(a):
    if len(a) == 1 and a.isalpha():
        return codes[ord(a.lower())-97]
    else:
        print("변환 오류 : 함수에 입력값이 여러개이거나 문자가 아닙니다.")
        return -1

if not text.isalpha():
    print("알파벳 문자열이 아닙니다.")
else:
    mores = ""
    for i in range(len(text)):
        mores = mores + translate(text[i]) + " "

    print(mores.rstrip())
```
