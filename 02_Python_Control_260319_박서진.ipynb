```python
# 세 개의 숫자 중 최댓값 구하기

# a, b, c = input("세 개의 숫자를 입력하세요 : ").split()
# a = int(a)
# b = int(b)
# c = int(c)
#print(type(a))

a, b, c = map(int, input("세 개의 숫자를 입력하세요 : ").split())

max_num = a

if max_num < b:
    max_num = b

if max_num < c:
    max_num = c

print(f"가장 큰 수는 {max_num}입니다.")
```

```python
# a > b 가 되도록 swap

a, b = map(int, input("두 수를 입력하세요 : ").split())
#print(">>", a, b)

if a < b:
    a, b = b, a

print(">>", a, b)
```

```python
# 나이의 차 구하기

age1 = int(input("첫번째 사람의 나이 : "))
age2 = int(input("두번째 사람의 나이 : "))

if age1 > age2:
    diff = age1 - age2
else:
    diff = age2 - age1

print(">> 나이 차이 :", diff)
```

```python
# 입력된 정수에 따라 등급 출력

score = int(input("점수를 입력하세요 : "))

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f">> 등급은 {grade}입니다.")
```

```python
#숫자를 입력 받아서 양수인 경우에만 짝수인지 홀수인지 출력한다. 양수가 아니면, 양수가 아니라고 출력

num = int(input())

if num > 0:
    if num % 2 == 0:
        print("짝수입니다.")
    else:
        print("홀수입니다.")
else:
    print("양수가 아닙니다")
```

```python
# while문 연습 : 나무 찍기

tree_hit = 0
while tree_hit < 10:
    tree_hit += 1    
    print(f"나무를 {tree_hit}번 찍었습니다.")

print("나무 넘어갑니다.")
```

```python
# while문 연습 : 입력값에 따라 반복 종료하기
# 사용자로부터 숫자를 하나씩 입력 받아서 총합과 평균을 출력
# 사용자가 0을 입력하면 숫자 입력을 멈춤

summary = count = 0
num = int(input("숫자 입력 (0 : 끝) :"))

while num != 0:
    summary += num
    count += 1
    num = int(input("숫자 입력 (0 : 끝) :"))

#0으로 나누기를 하면 ZeroDivisionError 발생 -> 예외처리
if count != 0:
    average = summary / count
else:
    average = 0

print(f">> 총합 : {summary}, 평균 : {average:.2f}")
```

```python
# 무한루프와 분기문 break 활용

summary = count = 0

while True:
    num = int(input("숫자 입력 (0 : 끝) :"))
    if num == 0:
        break
    summary += num
    count += 1

#0으로 나누기를 하면 ZeroDivisionError 발생 -> 예외처리
if count != 0:
    average = summary / count
else:
    average = 0

print(f">> 총합 : {summary}, 평균 : {average:.2f}")
```

```python
# for문 연습

summary = 0

for i in 1.1, 2.5, 3.6, 4.2, 5.4:
    summary += i
    print(i, summary)

print(f"합 : {summary}, 평균 : {summary / 5:.2f}")
```

```python
# 5개의 숫자를 계속 입력 받아서 모두 더한 값을 출력

summary = 0
print("5개의 숫자를 입력하세요.")

for _ in range(5):
    num = int(input())
    summary += num

print(f"5개 숫자의 합은 {summary}입니다.")
```

```python
# 숫자를 입력 받아서 1부터 해당 숫자까지 더하여 출력

num = int(input("숫자를 입력하세요 :"))
summary = 0

for i in range(1, num+1):
    summary += i

print(f"1부터 {num}까지의 합은 {summary}입니다.")
```

```python
# 입력 받은 숫자에 해당하는 구구단 출력

num = int(input("숫자를 입력하세요 :"))
print(f"[ 구구단 {num}단 ]")

for i in range(1, 10):
    print(f"{num} x {i} = {num*i}")
```

```python
# 입력받은 숫자의 팩토리얼 값을 출력

num = int(input())
summary = 1

if num != 0:
    for i in range(1, num+1):
        summary *= i
    
print(f"{num}! = {summary}")
```

```python
# 2단부터 9단까지의 구구단을 모두 출력

for num in range(2, 10):
    # num에 해당하는 구구단 출력
    print(f"[ 구구단 {num}단 ]")

    for i in range(1, 10):
        print(f"{num} x {i} = {num*i}")
    print()
```
