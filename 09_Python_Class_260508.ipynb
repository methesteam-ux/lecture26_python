### 파이썬 클래스

```python
class Circle:
    pi = 3.14
    
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return self.radius ** 2 * Circle.pi

c = Circle(2.5)
print(c.area())
```

### 사각형 넓이 구하는 클래스

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
    
rec = Rectangle(2.1, 3.4)
print(rec.area())
```

### 은행 클래스 예제

```python
class Account:
    def __init__(self):
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

a1 = Account()
a1.deposit(100)
print(a1.balance)
```

```python
class Account:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

a2 = Account()
a2.deposit(50)
print(a2.get_balance())
```

### 매직 메서드 연습

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}({self.age}살)"
    
    def __eq__(self, other):
        if self.name == other.name and self.age == other.age:
            return True
        return False

p1 = Person("홍길동", 15)
print(p1)

p2 = Person("홍길동", 15)
print(p2)

if p1 == p2:
    print("같은 사람입니다.")
else:
    print("다른 사람입니다.")
```

### 학습 내용 되돌아보기 (2)

```python
class Integer:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
    
    def __sub__(self, other):
        return self.value - other.value
    
a = Integer(5)
b = Integer(3)

print(a + b) #8
print(a - b) #2
```

### 학습 내용 되돌아보기 (4)

```python
class Student:
    def __init__(self, name, dept, mid, final):
        self.name = name
        self.dept = dept
        self.mid = mid
        self.final = final

    def __str__(self):
        return "학과: " + self.dept + ", 이름: " + self.name + ", 중간: " + str(self.mid) + ", 기말: " + str(self.final)
    
    def grade(self):
        if (self.mid + self.final) / 2 >= 90:
            return "A"
        elif (self.mid + self.final) / 2 >= 80:
            return "B"
        elif (self.mid + self.final) / 2 >= 70:
            return "C"
        elif (self.mid + self.final) / 2 >= 60:
            return "D"
        else:
            return "F"
        
student1 = Student("김경철", "기계학과", 89, 90)
print(student1)
print("학점: " + student1.grade())
```

### 학습 내용 되돌아보기 (5)

```python
class Car:
    def __init__(self, company, madewhen, color):
        self.company = company
        self.madewhen = madewhen
        self.color = color

    def __str__(self):
        return "자동차 회사: " + self.company + ", 년식: " + str(self.madewhen) + ", 색상: " + self.color
    
    def __eq__(self, other):
        if self.company == other.company and self.madewhen == other.madewhen and self.color == other.color:
            return True
        else:
            return False
        
mycar = Car("현대", 2020, "검정")
yourcar = Car("기아", 2021, "백색")
print(mycar)
print(yourcar)
print(mycar == yourcar)
```
