```python
s1 = {1, 2, 3}
print(s1)

s2 = set([1, 2, 3])
print(s2)

s3 = {"a", "b", "c"}
print(s3)

s4 = set("abc")
print(s4)

s5 = set("Hello")
print(s5)

try:
    s6 = set([1, [2, 3]])
    print(s6)
except:
    print("수정 가능한 리스트는 집합의 원소로 사용 불가")
```

```python
s1 = {1, 2, 3}
print(s1)

s1.add(4)
print(s1)

s1.update([5, 6, 7])
print(s1)

s1.remove(2)
print(s1)

try:
    s1.remove(8)
except:
    print("집한 s1 안에 원소 '8'이 존재하지 않음")
s1.discard(8)
print(s1)

item = s1.pop()
print(s1, item)

s1.clear()
print(s1)
```

```python
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])

s_inter1 = s1 & s2
s_inter2 = s1.intersection(s2)
print(s_inter1)
print(s_inter2)

s_union1 = s1 | s2
s_union2 = s1.union(s2)
print(s_union1)
print(s_union2)

s_diff1 = s1 - s2
s_diff2 = s1.difference(s2)
print(s_diff1)
print(s_diff2)

s_symdiff1 = s1 ^ s2
s_symdiff2 = s1.symmetric_difference(s2)
print(s_symdiff1)
print(s_symdiff2)
```

```python
S = input().strip()
result = set()

for i in range(len(S)): #부분 문자열의 시작 인덱스
    for j in range(i, len(S)): #부분 문자열의 스탑 인덱스
        sub_str = S[i:j+1]
        result.add(sub_str)

print(len(result))
```

```python
import sys
N = int(input())
in_person = set()

for _ in range(N):
    name, action = input().split()
    if action == 'enter':
        in_person.add(name)
    else:
        in_person.remove(name)

print("\n".join(sorted(in_person, reverse=True)))
```
